```python

import pandas as pd

# Step 1: Create the training dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Step 2: Display the training dataset
print("TRAINING DATASET:")
print(df.to_string(index=False))

# Step 3: Initialize the most specific hypothesis
hypothesis = ['0'] * (len(df.columns) - 1)
attributes = list(df.columns[:-1])

# Step 4: Apply FIND-S algorithm
for _, row in df.iterrows():
    if row['EnjoySport'] == 'Yes':
        for i, attribute in enumerate(attributes):
            if hypothesis[i] == '0':
                hypothesis[i] = row[attribute]
            elif hypothesis[i] != row[attribute]:
                hypothesis[i] = '?'

# Step 5: Display the final hypothesis
print("\nFINAL HYPOTHESIS:")
print(hypothesis)

# Display hypothesis with attribute names
print("\nHYPOTHESIS WITH ATTRIBUTES:")
for attribute, value in zip(attributes, hypothesis):
    print(f"{attribute}: {value}")
```
